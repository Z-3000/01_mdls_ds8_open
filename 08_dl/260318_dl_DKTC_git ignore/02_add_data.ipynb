```python
def show_structure(base_path, max_depth=10):
    for root, dirs, files in os.walk(base_path):
        depth = root.replace(base_path, "").count(os.sep)
        if depth > max_depth:
            dirs.clear()  # 더 깊은 탐색도 차단 (성능)
            continue
        indent = "  " * depth
        folder_name = os.path.basename(root) or base_path
        print(f"{indent}{folder_name}/")

        # 최대 깊이이거나 하위 디렉토리 없는 말단 폴더면 파일 요약
        if depth == max_depth or not dirs:
            if files:
                exts = {}
                for f in files:
                    ext = os.path.splitext(f)[1] or "(확장자없음)"
                    exts[ext] = exts.get(ext, 0) + 1
                print(f"{indent}  → 파일: {exts}")


base = r"/home/jovyan/d/git/01_mdls_ds8_open/08_dl/260318_dl_DKTC/00_data"

datasets = [
    "020.주제별 텍스트 일상 대화 데이터",
    "044.페르소나 대화",
    "011.일상대화 한국어 멀티세션 데이터",
]

for dataset in datasets:
    path = os.path.join(base, dataset)
    print(f"\n{'='*50}")
    print(f"데이터셋: {dataset}")
    print(f"경로 존재: {os.path.exists(path)}")
    show_structure(path)
```

```python
import os
import zipfile
from pathlib import Path


def _sanitize_member_path(member_name: str) -> str | None:
    """
    zip 멤버 경로를 정규화.
    - 선행 '/', '../' 제거
    - 결과가 빈 문자열이면 None 반환 (스킵 대상)

    예시:
      '../TS_가족/file.json'  → 'TS_가족/file.json'
      '../../etc/passwd'      → 'etc/passwd'  (깊은 탈출도 strip)
      '/absolute/path.json'   → 'absolute/path.json'
      'normal/path.json'      → 'normal/path.json'
    """
    # PureWindowsPath로도 처리해야 Windows에서 만든 zip도 커버
    parts = Path(member_name).parts

    # 위험 요소 제거: '/', '..', 드라이브 문자 등
    cleaned = []
    for p in parts:
        if p in ("..", "/", "\\") or (len(p) == 2 and p[1] == ":"):
            continue
        cleaned.append(p)

    if not cleaned:
        return None

    return os.path.join(*cleaned)


def safe_unzip_all(
    base_path: str,
    dataset_dir: str,
    max_print: int = 5,
    dry_run: bool = False,
):
    """
    base_path/dataset_dir 아래의 모든 .zip을 찾아 안전하게 해제.

    개선 사항 (v3):
    - Zip Slip 경로를 차단 대신 정규화하여 안전 추출
    - 멤버별 개별 추출 (extractall 대신)
    - dry_run=True 시 실제 해제 없이 진단만 출력
    - 정규화된 경로 수를 별도 카운팅
    """
    target = os.path.join(base_path, dataset_dir)

    if not os.path.isdir(target):
        print(f"[오류] 대상 디렉토리 없음: {target}")
        return

    # ── zip 파일 수집 ──
    zip_files = []
    for root, dirs, files in os.walk(target):
        for f in files:
            if f.lower().endswith(".zip"):
                zip_files.append(os.path.join(root, f))

    print(f"[{dataset_dir}] 찾은 zip: {len(zip_files)}개")
    if dry_run:
        print("  *** DRY RUN 모드 — 실제 해제 없음 ***")

    # ── 카운터 & 로그 ──
    count = {
        "skipped_empty": 0,
        "skipped_done": 0,
        "extracted": 0,
        "failed": 0,
        "sanitized_paths": 0,  # 경로 정규화가 필요했던 멤버 수
    }
    failures = []

    for i, zf in enumerate(zip_files):
        extract_to = os.path.dirname(zf)
        label = f"[{i+1}/{len(zip_files)}]"

        try:
            with zipfile.ZipFile(zf, "r") as z:
                members = z.namelist()

                # 1) 빈 zip 스킵
                if not members:
                    count["skipped_empty"] += 1
                    if i < max_print:
                        print(f"{label} 스킵(빈 zip): {os.path.basename(zf)}")
                    continue

                # 2) 경로 정규화 맵 구축: {원본경로: 정규화경로}
                path_map = {}
                sanitized_count = 0
                for m in members:
                    clean = _sanitize_member_path(m)
                    if clean is None:
                        continue
                    if clean != m:
                        sanitized_count += 1
                    path_map[m] = clean

                # 3) 실제 파일 멤버만 (디렉토리 제외)
                file_members = {
                    orig: clean
                    for orig, clean in path_map.items()
                    if not orig.endswith("/")
                }

                if not file_members:
                    count["skipped_empty"] += 1
                    if i < max_print:
                        print(f"{label} 스킵(파일 없음): {os.path.basename(zf)}")
                    continue

                # 4) 이미 풀렸는지 판단 — 정규화 경로 기준 80% 이상 존재
                existing = sum(
                    1 for clean in file_members.values()
                    if os.path.exists(os.path.join(extract_to, clean))
                )
                ratio = existing / len(file_members)

                if ratio >= 0.8:
                    count["skipped_done"] += 1
                    if i < max_print:
                        print(
                            f"{label} 스킵(이미 풀림 "
                            f"{existing}/{len(file_members)}): "
                            f"{os.path.basename(zf)}"
                        )
                    continue

                # 5) 멤버별 안전 추출
                if dry_run:
                    if i < max_print:
                        print(
                            f"{label} [DRY] 해제 예정 "
                            f"({len(file_members)}개 파일, "
                            f"정규화 {sanitized_count}개): "
                            f"{os.path.basename(zf)}"
                        )
                        # 정규화된 경로 샘플 출력
                        if sanitized_count > 0:
                            for orig, clean in list(path_map.items())[:3]:
                                if orig != clean:
                                    print(f"       {orig}  →  {clean}")
                            if sanitized_count > 3:
                                print(f"       ... +{sanitized_count - 3}개")
                else:
                    extracted_files = 0
                    for orig, clean in path_map.items():
                        info = z.getinfo(orig)
                        dest = os.path.join(extract_to, clean)

                        # 디렉토리 엔트리는 생성만
                        if orig.endswith("/"):
                            os.makedirs(dest, exist_ok=True)
                            continue

                        # 파일: 부모 디렉토리 확보 후 추출
                        os.makedirs(os.path.dirname(dest), exist_ok=True)
                        with z.open(orig) as src, open(dest, "wb") as dst:
                            dst.write(src.read())
                        extracted_files += 1

                    count["extracted"] += 1
                    count["sanitized_paths"] += sanitized_count

                    if i < max_print:
                        extra = f", 경로정규화 {sanitized_count}개" if sanitized_count else ""
                        print(
                            f"{label} 완료({extracted_files}개 파일{extra}): "
                            f"{os.path.basename(zf)}"
                        )

        except zipfile.BadZipFile as e:
            count["failed"] += 1
            failures.append((os.path.basename(zf), f"손상된 zip: {e}"))
            if i < max_print:
                print(f"{label} 실패(손상): {os.path.basename(zf)}")

        except Exception as e:
            count["failed"] += 1
            failures.append((os.path.basename(zf), str(e)))
            if i < max_print:
                print(f"{label} 실패: {os.path.basename(zf)} → {e}")

    # ── 요약 ──
    print(f"\n[{dataset_dir}] 처리 요약")
    print(f"  해제 완료   : {count['extracted']}개")
    print(f"  스킵(완료)  : {count['skipped_done']}개")
    print(f"  스킵(빈것)  : {count['skipped_empty']}개")
    print(f"  실패        : {count['failed']}개")
    if count["sanitized_paths"]:
        print(f"  경로 정규화  : {count['sanitized_paths']}개 멤버")

    if failures:
        print(f"\n  ── 실패 목록 ──")
        for fname, reason in failures:
            print(f"    {fname}: {reason}")

    print()


# ── 실행 ──
if __name__ == "__main__":
    base = "/home/jovyan/d/git/01_mdls_ds8_open/08_dl/260318_dl_DKTC/00_data"

    datasets = [
        "020.주제별 텍스트 일상 대화 데이터",
        "044.페르소나 대화",
        "011.일상대화 한국어 멀티세션 데이터",
    ]

    # 1단계: dry_run=True로 먼저 확인
    for ds in datasets:
        safe_unzip_all(base, ds, dry_run=True)

    # 2단계: 확인 후 실제 해제
    # for ds in datasets:
    #     safe_unzip_all(base, ds)
```

# 02_JSON 구조 확인

```python
import os
import json
from collections import defaultdict


# 유틸리티
def _schema(obj, depth: int = 0, max_depth: int = 4):
    """
    JSON 객체의 스키마를 재귀적으로 추출.
    반환 형태 예시: {"key1": "str", "key2": ["{'turn': 'str', 'text': 'str'}"]}
    """
    if depth >= max_depth:
        return "..."

    if isinstance(obj, dict):
        return {k: _schema(v, depth + 1, max_depth) for k, v in obj.items()}
    elif isinstance(obj, list):
        if not obj:
            return ["empty"]
        # 첫 항목의 스키마만 대표로 추출
        return [_schema(obj[0], depth + 1, max_depth)]
    else:
        return type(obj).__name__


def _flat_keys(obj, prefix: str = ""):
    """
    중첩 dict/list를 '.' 구분 플랫 키로 변환.
    예: {"info": {"topic": "..."}} → {"info.topic"}
    리스트는 [0]으로 한 단계만 진입.
    """
    keys = set()
    if isinstance(obj, dict):
        for k, v in obj.items():
            full = f"{prefix}.{k}" if prefix else k
            keys.add(full)
            keys |= _flat_keys(v, full)
    elif isinstance(obj, list) and obj and isinstance(obj[0], (dict, list)):
        keys |= _flat_keys(obj[0], f"{prefix}[]")
    return keys


def _sample_text(obj, char_limit: int = 600):
    """
    JSON 객체를 보기 좋게 출력하되, 리스트는 첫 2개 항목만,
    문자열 값은 80자로 잘라서 전체 구조가 보이도록 함.
    """
    def _trim(o, depth=0):
        if depth > 5:
            return "..."
        if isinstance(o, dict):
            return {k: _trim(v, depth + 1) for k, v in o.items()}
        elif isinstance(o, list):
            trimmed = [_trim(item, depth + 1) for item in o[:2]]
            if len(o) > 2:
                trimmed.append(f"... +{len(o)-2}개")
            return trimmed
        elif isinstance(o, str) and len(o) > 80:
            return o[:77] + "..."
        return o

    text = json.dumps(_trim(obj), ensure_ascii=False, indent=2)
    if len(text) > char_limit:
        return text[:char_limit] + "\n  ... (truncated)"
    return text



# 메인 함수
def inspect_json_structure(
    base_path: str,
    dataset_dir: str,
    samples_per_folder: int = 1,
    max_folders: int = 5,
):
    """
    base_path/dataset_dir 아래 JSON 파일의 구조를 폴더별로 탐색.

    - 폴더별로 samples_per_folder개씩 샘플링 (스키마가 폴더마다 다를 수 있으므로)
    - 스키마 재귀 추출 + 플랫 키 목록
    - 전체 요약: 고유 스키마 수, 공통/차이 키
    """
    target = os.path.join(base_path, dataset_dir)

    if not os.path.isdir(target):
        print(f"[오류] 디렉토리 없음: {target}")
        return {}

    # 폴더별 JSON 수집
    folder_map = defaultdict(list)  # {folder_path: [json_paths]}
    for root, dirs, files in os.walk(target):
        for f in files:
            if f.lower().endswith(".json"):
                folder_map[root].append(os.path.join(root, f))

    total_files = sum(len(v) for v in folder_map.values())
    print("=" * 60)
    print(f"[{dataset_dir}]")
    print(f"  JSON 파일 수: {total_files}개  |  폴더 수: {len(folder_map)}개")
    print("=" * 60)

    if not folder_map:
        print("  → JSON 파일 없음\n")
        return {}

    # 폴더별 샘플링 & 스키마 추출
    all_schemas = []       # (folder_short, schema_dict, flat_keys)
    folder_items = sorted(folder_map.items())

    for fi, (folder, json_list) in enumerate(folder_items):
        if fi >= max_folders:
            remaining = len(folder_items) - max_folders
            print(f"\n  ... +{remaining}개 폴더 생략 (max_folders={max_folders})")
            break

        rel_folder = os.path.relpath(folder, target)
        print(f"\n── 폴더: {rel_folder}  ({len(json_list)}개 파일) ──")

        for si in range(min(samples_per_folder, len(json_list))):
            jf = json_list[si]
            try:
                with open(jf, "r", encoding="utf-8") as f:
                    data = json.load(f)
            except (json.JSONDecodeError, UnicodeDecodeError) as e:
                print(f"  [읽기 실패] {os.path.basename(jf)}: {e}")
                continue

            top_type = type(data).__name__
            sample = data[0] if isinstance(data, list) and data else data

            schema = _schema(sample)
            flat = _flat_keys(sample)
            all_schemas.append((rel_folder, schema, flat))

            print(f"  파일: {os.path.basename(jf)}")
            print(f"  최상위 타입: {top_type}", end="")
            if isinstance(data, list):
                print(f"  |  길이: {len(data)}", end="")
            print()

            print(f"  스키마:")
            schema_str = json.dumps(schema, ensure_ascii=False, indent=4)
            for line in schema_str.split("\n"):
                print(f"    {line}")

            print(f"  샘플:")
            for line in _sample_text(sample).split("\n"):
                print(f"    {line}")

    # 데이터셋 내 스키마 일관성 요약
    if len(all_schemas) > 1:
        print(f"\n── 스키마 일관성 검사 ──")
        key_sets = [s[2] for s in all_schemas]
        common = set.intersection(*key_sets)
        all_keys = set.union(*key_sets)
        diff = all_keys - common

        print(f"  공통 키 ({len(common)}개): {sorted(common)}")
        if diff:
            print(f"  차이 키 ({len(diff)}개): {sorted(diff)}")
            for folder, _, flat in all_schemas:
                unique = flat - common
                if unique:
                    print(f"    {folder} 고유: {sorted(unique)}")
        else:
            print(f"  → 모든 폴더 스키마 동일")

    print()
    return {s[0]: s[2] for s in all_schemas}



# 실행 + 크로스 데이터셋 비교
if __name__ == "__main__":
    base = "/home/jovyan/d/git/01_mdls_ds8_open/08_dl/260318_dl_DKTC/00_data"

    datasets = [
        "020.주제별 텍스트 일상 대화 데이터",
        "044.페르소나 대화",
        "011.일상대화 한국어 멀티세션 데이터",
    ]

    dataset_keys = {}  # {dataset_dir: set(all_flat_keys)}

    for ds in datasets:
        result = inspect_json_structure(base, ds, samples_per_folder=1, max_folders=5)
        if result:
            merged = set()
            for keys in result.values():
                merged |= keys
            dataset_keys[ds] = merged

    # 크로스 데이터셋 비교
    if len(dataset_keys) > 1:
        print("=" * 60)
        print("크로스 데이터셋 스키마 비교")
        print("=" * 60)

        all_ds_keys = list(dataset_keys.values())
        cross_common = set.intersection(*all_ds_keys)
        cross_all = set.union(*all_ds_keys)

        print(f"\n  전체 데이터셋 공통 키: {sorted(cross_common) if cross_common else '없음'}")
        print(f"  전체 고유 키 수: {len(cross_all)}개\n")

        for ds, keys in dataset_keys.items():
            unique = keys - cross_common
            ds_short = os.path.basename(ds) if os.sep in ds else ds
            print(f"  [{ds_short}]")
            print(f"    전체 키 수: {len(keys)}개")
            if unique:
                print(f"    고유 키: {sorted(unique)}")
            print()

        print("  → 공통 키가 없으면 데이터셋별 매핑 함수가 필요합니다.")
        print("    (대화 텍스트 필드, 주제 레이블 필드를 각각 식별 후 통합)")
        print()
```

## 011_발화단위 csv 파일 만들기

```python
import os
import json
from pathlib import Path
from typing import Dict, Any, List, Optional

import pandas as pd


BASE_COLUMNS = [
    "dataset",
    "doc_id",
    "session_id",
    "turn_id",
    "speaker_id",
    "speaker_role",
    "text",
    "topic_raw",
    "topic_large",
    "topic_medium",
    "topic_small",
    "category",
    "source_file",
]

EXTRA_COLUMNS = [
    "speaker_age",
    "speaker_gender",
    "speaker_occupation",
    "session_summary",
    "prev_session_id",
    "prev_aggregated_summary",
    "context_query",
    "context_document_title",
]

ALL_COLUMNS = BASE_COLUMNS + EXTRA_COLUMNS


def safe_get(d: Any, *keys, default=None):
    """
    중첩 dict 안전 접근
    """
    cur = d
    for k in keys:
        if not isinstance(cur, dict) or k not in cur:
            return default
        cur = cur[k]
    return cur


def json_dumps_safe(obj: Any) -> Optional[str]:
    """
    dict/list를 CSV 저장용 문자열로 변환
    """
    if obj is None:
        return None
    try:
        return json.dumps(obj, ensure_ascii=False)
    except Exception:
        return str(obj)


def build_participant_lookup(participants_info: Dict[str, Any]) -> Dict[str, Dict[str, Any]]:
    """
    participantsInfo 안의 speaker1, speaker2를 participantID 기준 lookup dict로 변환
    """
    lookup = {}

    for speaker_key in ["speaker1", "speaker2"]:
        sp = participants_info.get(speaker_key, {})
        participant_id = sp.get("participantID")
        if participant_id:
            lookup[participant_id] = {
                "speaker_age": sp.get("age"),
                "speaker_gender": sp.get("gender"),
                "speaker_occupation": sp.get("occupation"),
            }

    return lookup


def extract_context_document_titles(dialog_item: Dict[str, Any]) -> Optional[str]:
    """
    sessionInfo[].dialog[].context.contents[].documentTitle 추출
    여러 개면 ' | ' 로 join
    """
    contents = safe_get(dialog_item, "context", "contents", default=[])
    if not isinstance(contents, list) or not contents:
        return None

    titles = []
    for c in contents:
        if isinstance(c, dict):
            title = c.get("documentTitle")
            if title:
                titles.append(str(title))

    if not titles:
        return None

    return " | ".join(titles)


def extract_011_rows_from_json(data: Dict[str, Any], source_file: str) -> List[Dict[str, Any]]:
    """
    011 JSON 1개에서 발화 단위 row 리스트 추출
    """
    rows = []

    file_info = data.get("FileInfo", {})
    participants_info = data.get("participantsInfo", {})
    topic_info = data.get("topicInfo", {})
    session_info_list = data.get("sessionInfo", [])

    participant_lookup = build_participant_lookup(participants_info)

    doc_id = file_info.get("filename", Path(source_file).stem)
    topic_large = topic_info.get("largeCategory")
    topic_medium = topic_info.get("mediumCategory")
    topic_small = topic_info.get("smallCategory")

    # topic_raw / category는 011에 딱 맞는 단일 필드가 없어서
    # 가장 안전하게 smallCategory를 topic_raw로 두고,
    # category는 largeCategory로 둠
    # 설계상 선택 사항
    topic_raw = topic_small
    category = topic_large

    if not isinstance(session_info_list, list):
        return rows

    for session in session_info_list:
        session_id = session.get("sessionID")
        prev_session_id = session.get("prevSessionID")
        session_summary = safe_get(session, "sessionSummary", "dialogSummary")
        prev_aggregated_summary = session.get("prevAggregatedSummary")
        dialog_list = session.get("dialog", [])

        if not isinstance(dialog_list, list):
            continue

        for idx, dialog in enumerate(dialog_list, start=1):
            speaker_id = dialog.get("participantID")
            speaker_role = dialog.get("speaker")
            text = dialog.get("utterance")

            # participantID 기반 화자 메타 lookup
            speaker_meta = participant_lookup.get(speaker_id, {})

            row = {
                "dataset": "011",
                "doc_id": doc_id,
                "session_id": session_id,
                "turn_id": idx,  # 원본에 명시 turn id가 없으므로 순번 사용
                "speaker_id": speaker_id,
                "speaker_role": speaker_role,
                "text": text,
                "topic_raw": topic_raw,
                "topic_large": topic_large,
                "topic_medium": topic_medium,
                "topic_small": topic_small,
                "category": category,
                "source_file": source_file,
                "speaker_age": speaker_meta.get("speaker_age"),
                "speaker_gender": speaker_meta.get("speaker_gender"),
                "speaker_occupation": speaker_meta.get("speaker_occupation"),
                "session_summary": session_summary,
                "prev_session_id": prev_session_id,
                "prev_aggregated_summary": json_dumps_safe(prev_aggregated_summary),
                "context_query": safe_get(dialog, "context", "query"),
                "context_document_title": extract_context_document_titles(dialog),
            }

            rows.append(row)

    return rows


def build_011_utterance_csv(
    input_root: str,
    output_csv_path: str,
    include_subdirs: Optional[List[str]] = None,
) -> pd.DataFrame:
    """
    011 데이터셋 전체 JSON을 발화 단위 CSV로 변환

    Parameters
    ----------
    input_root : str
        예: /home/jovyan/.../011.일상대화 한국어 멀티세션 데이터/
    output_csv_path : str
        저장할 csv 경로
    include_subdirs : list[str] | None
        예: ["Training", "Validation"]
        None이면 input_root 하위 전체 탐색
    """
    input_root = Path(input_root)
    all_rows = []

    if not input_root.exists():
        raise FileNotFoundError(f"입력 경로가 없습니다: {input_root}")

    if include_subdirs:
        json_files = []
        for subdir in include_subdirs:
            sub_path = input_root / subdir
            if sub_path.exists():
                json_files.extend(sorted(sub_path.rglob("*.json")))
    else:
        json_files = sorted(input_root.rglob("*.json"))

    print(f"JSON 파일 수: {len(json_files)}")

    for i, json_path in enumerate(json_files, start=1):
        try:
            with open(json_path, "r", encoding="utf-8") as f:
                data = json.load(f)

            rel_path = str(json_path.relative_to(input_root))
            rows = extract_011_rows_from_json(data, source_file=rel_path)
            all_rows.extend(rows)

        except Exception as e:
            print(f"[오류] {json_path}: {e}")

        if i % 1000 == 0:
            print(f"{i:,} / {len(json_files):,} 처리 완료")

    df = pd.DataFrame(all_rows)

    # 누락 컬럼 강제 생성
    for col in ALL_COLUMNS:
        if col not in df.columns:
            df[col] = None

    # 컬럼 순서 고정
    df = df[ALL_COLUMNS]

    # 저장
    output_csv_path = Path(output_csv_path)
    output_csv_path.parent.mkdir(parents=True, exist_ok=True)
    df.to_csv(output_csv_path, index=False, encoding="utf-8-sig")

    print(f"저장 완료: {output_csv_path}")
    print(f"shape: {df.shape}")

    return df


#  실행 
df_011_train = build_011_utterance_csv(
    input_root="/home/jovyan/d/git/01_mdls_ds8_open/08_dl/260318_dl_DKTC/00_data/011.일상대화 한국어 멀티세션 데이터/Training",
    output_csv_path="/home/jovyan/d/git/01_mdls_ds8_open/08_dl/260318_dl_DKTC/00_data_csv/011_train_utterances.csv",
    include_subdirs=None,
)

df_011_val = build_011_utterance_csv(
    input_root="/home/jovyan/d/git/01_mdls_ds8_open/08_dl/260318_dl_DKTC/00_data/011.일상대화 한국어 멀티세션 데이터/Validation",
    output_csv_path="/home/jovyan/d/git/01_mdls_ds8_open/08_dl/260318_dl_DKTC/00_data_csv/011_val_utterances.csv",
    include_subdirs=None,
)
```
